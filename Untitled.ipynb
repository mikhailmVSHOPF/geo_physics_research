```python
import matplotlib.pyplot as plt 
import seaborn as sns 
import numpy as np
```

```python

x = 0.14
y = 0.1
R = np.sqrt((1+x)**2 + y**2)
phi = np.linspace(0, 2*np.pi, 1000)
z = R*np.exp(phi*1j) + y*1j + x
w = 0.5 * (z + 1/z)

plt.figure(figsize=(12, 4))
plt.scatter(w.real, w.imag, s = 1)
plt.axis('equal')
plt.grid()
plt.show()
```

```python

y = np.linspace(min(w.imag)-1, max(w.imag)+1, 1000)
x = np.linspace(min(w.real)-1, max(w.real)+1, 1000)

X, Y = np.meshgrid(x,y)
```
